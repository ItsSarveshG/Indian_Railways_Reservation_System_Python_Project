```python
# Create Fare Calculation Function
```

```python
def calculate_fare(ticket_price, passengers):
  total_amount = ticket_price * passengers
  return total_amount
```

```python
# Call the Function
```

```python
final_fare = calculate_fare(2450, 2)
print("Final Fare :", final_fare)
```

```python
# Printing Factorial of a Number using Recursive Function

def fact(n):
  if n ==0 or n ==1:
    return 1
  return n * fact(n - 1)

num = int(input("Enter a Number: "))
print("Factorial:", fact(num))
```
