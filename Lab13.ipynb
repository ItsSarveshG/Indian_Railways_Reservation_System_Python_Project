```python
# Task 6: Process Data Using map() and filter()

ticket_fares = [2500, 800, 3200, 600]

valid_fares = list(
    filter(lambda fare : fare> 1000, ticket_fares))
```

```python
final_fares = list(
    map(
        lambda fare: fare + 100,
        valid_fares))

print(final_fares)
```

```python
# Task 7: Build Final Functional Reservation Module

passengers = ["Rahul", "Priya", "Amit"]
seats = ["S1", "S2", "S3"]
fares = [2500, 1800, 3200]

reservation_data = list(zip(passengers, seats, fares))
```

```python
print("==========================")
print("Railway Reservation Report")
print("==========================")

for index, data in enumerate(reservation_data):

  passenger, seat, fare = data

  final_fare = (lambda x: x - 100) (fare)

  print("Reservation ID:", index + 1)
  print("Passenger:", passenger)
  print("Seat:", seat)
  print("Final Fare:", final_fare)
  print("=========================")
```

```python
# How to use map function & make it in list
l1 = [4,9,6,2]
l2 = list(map(lambda num:num*num,l1))

print(l2)
```

```python
# How to use map function & without making it list
l1 = [4,9,6,2]
l2 = map(lambda num:num*num,l1)

print(l2)
```

```python
# lambda      👉 one line anonymous function
# map()       👉 apply function on each list members
# filter()    👉 Filters data from list based on condition
# zip()       👉 Combines two or more list members
# enumerate() 👉 returns Index + Value
# list()      👉 Converts data into list (type casting)
# print()     👉 Displays the output on console
# IIFE        👉 Immediately Invoked Functions Expression
```
